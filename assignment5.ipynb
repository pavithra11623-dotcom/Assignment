Use bar chart or pie chart to display the frequency of categories for categorical variables.

```python
#Here we cant plot the bar chart for all the values as dataset has the many different unique values(>20)
#That's the reason here we ploted top 12 values
for col in data:
    plt.figure(figsize=(6,4))
    data[col].value_counts().head(12).plot(kind='bar', color='blue', edgecolor='black')
    plt.title(f"Frequency of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```
