```python
X = df.drop('Y', axis=1)
y = df['Y']
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
le = LabelEncoder()
y_encoded = le.fit_transform(y)
```

```python
print("Preprocessing Complete!")
print("X_scaled Shape:", X_scaled.shape)
print("y_encoded Shape:", y_encoded.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_encoded, test_size=0.2, random_state=42)
```

2. Model Implementation

●	Construct a basic ANN model using your chosen high-level neural network library. Ensure your model includes at least one hidden layer.

●	Divide the dataset into training and test sets.

●	Train your model on the training set and then use it to make predictions on the test set.


```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
import warnings
warnings .filterwarnings("ignore")
```

```python
model = Sequential()
model.add(Dense(64, activation='relu', input_shape=(X_train.shape[1],)))
model.add(Dense(32, activation='relu'))
model.add(Dense(len(set(y_encoded)), activation='softmax'))
```

```python
model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=50, batch_size=32, validation_data=(X_test, y_test))
```

3. Hyperparameter Tuning

●	Modify various hyperparameters, such as the number of hidden layers, neurons per hidden layer, activation functions, and learning rate, to observe their impact on model performance.

●	Adopt a structured approach like grid search or random search for hyperparameter tuning, documenting your methodology thoroughly.


```python
from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import GridSearchCV
```

```python
def create_model(activation='relu', neurons=64, optimizer='adam'):
    model = Sequential()
    model.add(Dense(neurons, activation=activation, input_shape=(X_train.shape[1],)))
    model.add(Dense(neurons//2, activation=activation))
    model.add(Dense(len(set(y_encoded)), activation='softmax'))
    model.compile(loss='sparse_categorical_crossentropy', optimizer=optimizer, metrics=['accuracy'])
    return model
```

```python
model = KerasClassifier(build_fn=create_model, verbose=0)
```

```python
param_grid = {
    'activation': ['relu', 'tanh'],
    'neurons': [32, 64, 128],
    'optimizer': ['adam', 'sgd']
}
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
```

```python

# Define the estimator (model)
estimator = RandomForestClassifier()
```

```python

# Define param_grid (example)
param_grid = {'n_estimators': [10, 50, 100]}
```

```python

# Run GridSearchCV
grid = GridSearchCV(estimator, param_grid, cv=3, scoring='accuracy')
```

```python
grid_result = grid.fit(X_train, y_train)
```

```python
print("Best Parameters:", grid_result.best_params_)
```

4. Evaluation

●	Employ suitable metrics such as accuracy, precision, recall, and F1-score to evaluate your model's performance.



```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print(y_pred)
```

```python
# Check shape of y_pred
print("y_pred shape:", y_pred.shape)
```

```python

# Fix argmax for multi-class (softmax output)
if len(y_pred.shape) > 1:
    y_pred_class = np.argmax(y_pred, axis=1)
else:
    y_pred_class = (y_pred > 0.5).astype(int)
```

```python
accuracy = accuracy_score(y_test, y_pred_class)
print(f"Accuracy: {accuracy:.4f}")
```

```python
precision = precision_score(y_test, y_pred_class, average='macro')
print(f"Precision (macro): {precision:.4f}")
```

```python
recall = recall_score(y_test, y_pred_class, average='macro')
print(f"Recall (macro): {recall:.4f}")
```

```python
f1 = f1_score(y_test, y_pred_class, average='macro')
print(f"F1‑Score (macro): {f1:.4f}")
```

●	Discuss the performance differences between the model with default hyperparameters and the tuned model, emphasizing the effects of hyperparameter tuning.

1. Default hyperparameters:

    - The model uses the out‑of‑the‑box settings (e.g., n_estimators=100, max_depth=None for RandomForest).

    - It may underfit or overfit depending on the dataset because the settings aren’t optimized for the specific problem.

    - Performance metrics (accuracy, precision, recall, F1) are usually moderate.

2. Tuned hyperparameters:

    - Tuning (via GridSearchCV, RandomSearch, etc.) searches for the best combo of parameters like n_estimators, max_depth, min_samples_split, etc.

    - A tuned model often reduces overfitting by controlling tree depth or increases model complexity to capture patterns better.

    - The tuned model typically shows improved metrics: higher accuracy, better precision/recall balance, and a boosted F1 score.

3. Effects of tuning:

    - Bias‑variance trade‑off: Tuning helps find a sweet spot, lowering variance (overfitting) or bias (underfitting).

    - Generalization: The tuned model usually generalizes better on unseen data, reflected in higher validation scores.
				
    - Computational cost: Tuning needs more computation (grid search can be expensive), but the performance gain often justifies it.
