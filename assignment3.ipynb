```python
data=data.drop('Date',axis=1) #dropping date column as it is not required.
print(data)
```

```python
data=data.drop('City',axis=1)# dropping city column as it is not required.
print(data)
```

```python
data_cat=data.select_dtypes(include='object')
print(data_cat)
```

```python
for col in data_cat:
    plt.figure(figsize=(7,5))
    data[col].value_counts().plot(kind='bar', color='red', edgecolor='black')
    plt.title(f"Frequency of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```
