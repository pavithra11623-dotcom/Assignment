6.	Given a 'marks' variable, print the grade:
- 90 and above: 'A'
- 75–89: 'B'
- 50–74: 'C'
- Below 50: 'Fail'


```python
m1 = int(input("enter your marks: "))
m2 = int(input("enter your marks: "))
c1 = int(input("enter your marks: "))
c2 = int(input("enter your marks: "))

total = sum([m1,m2,c1,c2])

avg = total/4
print(f"your avg marks of all subjects is {avg}")
if avg>90:
    print(f"Congratulations your grade is A")
elif avg>89 and avg<=75:
    print(f"Congratulations your grade is B")
elif avg>74 and avg<=50:
    print(f"can do better you got C")
elif avg>50:
    print(f"can do more better you got fail")
```

11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows


```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Hannah', 'Ivy', 'Jack'],
    'Age': [25, 32, 18, 47, 35, 29, 41, 23, 30, 52],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 'New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
    'Salary': [50000, 65000, np.nan, 90000, 75000, 52000, np.nan, 48000, 72000, 95000]
}
df = pd.DataFrame(data)
print("DataFrame:")
print(df)
```

```python
df.info()
```

```python
print(df.describe())
```

```python
selected_columns = df[['Name', 'City']]
print(selected_columns)
```

```python
df = df.drop(columns=['City'])
print(df)
```

```python
salary_mean = df['Salary']
print(salary_mean)
```

```python
df.drop_duplicates(inplace=True)
print("\nDataFrame after dropping duplicates:")
print(df)
```
