Task 2: Dimensionality Reduction with PCA:

1.	Standardize the features to ensure they have a mean of 0 and a standard deviation of Implement PCA to reduce the dimensionality of the dataset. 

2.	Determine the optimal number of principal components using techniques like scree plot or cumulative explained variance.

3.	Transform the original dataset into the principal components.


```python
from sklearn.preprocessing	import StandardScaler
from sklearn.decomposition	import PCA
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(Wine.drop('Alcohol', axis=1))
```

```python
pca = PCA()
pca.fit(X_scaled)
```

```python
print(pca.explained_variance_ratio_)
```

```python
plt.figure(figsize=(8,6))
plt.plot(pca.explained_variance_ratio_, marker='o')
plt.title("Scree Plot")
plt.xlabel("Principal Components")
plt.ylabel("Explained Variance Ratio")
plt.show()
```

```python
cum_var = np.cumsum(pca.explained_variance_ratio_)
plt.figure(figsize=(8,6))
plt.plot(cum_var, marker='o')
plt.title("Cumulative Explained Variance")
plt.xlabel("Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.show()
```

```python
optimal_components = np.where(cum_var >= 0.9)[0][0] + 1
print(f"Optimal number of principal components: {optimal_components}")
```

```python
pca_optimal = PCA(n_components=optimal_components)
X_pca = pca_optimal.fit_transform(X_scaled)
```

Task 3: Clustering with Original Data:

1.	Apply a clustering algorithm (e.g., K-means) to the original dataset.

2.	Visualize the clustering results using appropriate plots.

3.	Evaluate the clustering performance using metrics such as silhouette score or Davies–Bouldin index.


```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score
```

```python
kmeans = KMeans(n_clusters=3)
kmeans.fit(Wine.drop('Alcohol', axis=1))
labels = kmeans.labels_
```

```python
# Silhouette score
score = silhouette_score(X_scaled, labels)
print('Silhouette Score:', score)
```

```python
pca = PCA(n_components=2)
df_pca = pca.fit_transform(X_scaled)
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x=Wine['Malic'], y=Wine['Ash'], hue=labels, palette='viridis')
plt.title("K-means Clustering Results")
plt.show()
```

```python
wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.plot(range(1, 11), wcss)
plt.title('Elbow Method')
plt.show()
```

```python
silhouette = silhouette_score(Wine.drop('Alcohol', axis=1), labels)
```

```python
print(f"Silhouette Score: {silhouette}")
```

```python
davies_bouldin = davies_bouldin_score(Wine.drop('Alcohol', axis=1), labels)
```

```python
print(f"Davies-Bouldin Index: {davies_bouldin}")
```

2.	Discuss any similarities or differences observed in the clustering results.

    - Similarities: Both methods may capture major patterns in the data, leading to some overlap in cluster structures, especially if the principal components retain most of the variance.

    - Differences: PCA reduces dimensionality, which can remove noise and lead to more distinct, meaningful clusters However, it may also lose some information, causing clusters to merge or shift. 
				
    - Visualizations (e.g., scatter plots) often show that PCA-based clustering produces more compact and separated clusters due to decorrelation of features.



3.	Reflect on the impact of dimensionality reduction on clustering performance.

      Noise Reduction: PCA removes less important variance (noise), which can improve clustering by focusing on significant patterns.

    - Information Loss: Discarding minor principal components may lose subtle distinctions between clusters, affecting performance.

    - Computational Efficiency: Clustering in reduced dimensions is faster and requires less memory.
				
    - Performance Metrics: Often, PCA improves metrics like silhouette score or Davies-Bouldin index by simplifying the feature space.



4.	Analyze the trade-offs between using PCA and clustering directly on the original dataset.

    - Advantages of PCA + Clustering:

        - Simplifies high-dimensional data.

        - Reduces overfitting by eliminating noise.

        - Speeds up clustering algorithms.

    - Disadvantages of PCA + Clustering:

        - Potential loss of discriminative information.

        - Requires choosing the number of components (variance retention).

        - Interpretation of principal components can be less intuitive than original features.

    - Direct Clustering on Original Data:

        - Retains all information, which may be useful for complex structures.

        - Computationally intensive and prone to noise in high dimensions.
								
        - May suffer from the curse of dimensionality.

3.	Provide recommendations for when to use each technique based on the analysis conducted.

    - Use PCA when:

        - Data has high dimensionality (e.g., images, gene expression).

        - Need to reduce noise or redundancy.

        - Visualizing data in 2D/3D is required.

    - Use Clustering when:

        - Exploring natural groupings in unlabeled data.

        - Identifying outliers or distinct subgroups.

        - Data is relatively low-dimensional or PCA has been applied first.

    - Combine PCA and Clustering:

        - Apply PCA to reduce noise, then cluster for better-defined groups.
								
        - Useful for large datasets where direct clustering is inefficient.
