Assignment Tasks
a. Build 99% Confidence Interval Using Sample Standard Deviation
Assuming the sample is representative of the population, construct a 99% confidence interval for the mean number of characters printed before the print-head fails using the sample standard deviation. Explain the steps you take and the rationale behind using the t-distribution for this task.


```python
import pandas as pd
import scipy
from scipy import stats
import numpy as np
```

```python
data = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
```

```python
n = len(data)
print(n)
```

```python
mean = np.mean(data)
print(mean)
```

```python
s = np.std(data, ddof=1)
print(s)
```

```python
alpha = 0.01
t_crit = stats.t.ppf(1-alpha/2, df=n-1)
margin_t = t_crit * (s / np.sqrt(n))
ci_t = (mean - margin_t, mean + margin_t)
print(ci_t)
```

```python
print("\n99% Confidence Interval(t-distribution:)")
print(f"{ci_t[0]:.4f} to {ci_t[1]:.4f}")
```

Build 99% Confidence Interval Using Known Population Standard Deviation
If it were known that the population standard deviation is 0.2 million characters, construct a 99% confidence interval for the mean number of characters printed before failure.


```python
sigma = 0.2
z_crit = stats.norm.ppf(1 - alpha/2)
print(z_crit)
```

```python
margin_z = z_crit * (sigma / np.sqrt(n))
print(margin_z)
```

```python
ci_z = (mean - margin_z,mean + margin_z)
print(ci_z)
```

```python
print("\n99% Confidence Interval(z-distribution, sigma = 0.2):")
print(f"{ci_z[0]:.4f} to {ci_z[1]:.4f}")
```

Calculate the Test Statistic:
Use the following formula to calculate the test statistic(z):
where:

```python
import math
from scipy.stats import norm
```

```python
x_bar = 3050 #sample mean
sigma = 525 #population standard deviation
n = 25 #sample size
```

```python
x = 600
mu_0 = 1000 + 55*x
print(mu_0)
```

```python
z = (x_bar - mu_0) / (sigma /math.sqrt(n))
print("\nCalculated Test Statistic (z):", round(z, 4))
```

Determine the Probability and compare:
Using the alpha level of 5% (α = 0.05),
Compare the test statistic with the critical value to decide whether to reject the null hypothesis

```python
alpha = 0.05
```

```python
p_value = 2*norm.cdf(z)
print("\n (alpha = 0.05):")
if p_value < alpha:
    print("Reject H0: The mean weekly cost is significantly different.")
else:
    print("Fail to Reject H0: No significant difference detected.")

```
